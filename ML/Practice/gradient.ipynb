```python
def gradient_descent(x, learning_rate, iterations):
    for i in range(iterations):
        gradient = 2 * (x + 3)
        x = x - learning_rate * gradient
        y = (x + 3) ** 2
        print(f"Iteration {i+1}: x = {x:.10f}, y = {y:.10f}")
    return x

x = 2
learning_rate = 0.1
iterations = 100

minimum_x = gradient_descent(x, learning_rate, iterations)

print("\nMinimum x:", minimum_x)
print("Minimum y:", (minimum_x + 3) ** 2)
```

```python
import numpy as np
import matplotlib.pyplot as plt
x = 2
learning_rate = 0.1
iterations = 100
start_x=x
x_values = [x]
y_values = [(x + 3)**2]

for i in range(iterations):
    gradient = 2 * (x + 3)
    x = x - learning_rate * gradient
    y=(x+3)**2
    print(f"Iteration {i+1}: x = {x:.10f}, y = {y:.10f}")
    x_values.append(x)
    y_values.append(y)

print("\n")
print("\nMinimum x:", x)
print("Minimum y:", (x + 3) ** 2)
# Plot function
x_axis = np.linspace(-5, 4, 200)
y_axis = (x_axis + 3)**2

plt.plot(x_axis, y_axis, label="y = (x + 3)^2")

# Plot Gradient Descent path
plt.scatter(x_values, y_values, color="red", label="Gradient Descent iterations")

plt.scatter(x, y, color="green", s=100, label="Minimum")
plt.scatter(start_x, (start_x+3)**2, color="orange", s=100, label="Starting point")

plt.xlabel("x")
plt.ylabel("y")
plt.title("Gradient Descent Visualization")
plt.legend()
plt.grid()
plt.show()
```
